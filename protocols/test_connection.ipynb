```python
from pylabrobot.liquid_handling.backends.tecan.EVO_backend import LiHa, RoMa
from pylabrobot.resources.tecan.wash import Wash_Station, Wash_Station_Cleaner_deep, Wash_Station_Cleaner_shallow, Wash_Station_Waste

lh = LiquidHandler(backend=EVOBackend(), deck=EVO200Deck())
lh.deck.get_resource(name='wash_station').unassign() # Assigned to rail 0 by default

wash_station = Wash_Station(name='wash station')
wash_station[0] = Wash_Station_Cleaner_deep("cleaner deep")
wash_station[1] = Wash_Station_Cleaner_shallow("cleaner shallow")
wash_station[2] = Wash_Station_Waste("waste")


lh.deck.assign_child_resource(wash_station, rails=38)

lh.summary()

# await lh.setup()
```
